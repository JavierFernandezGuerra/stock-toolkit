# Exploratory Analysis

Este cuaderno ejemplifica cómo cargar datos normalizados del toolkit y explorar estadísticos descriptivos. Completa los pasos siguientes y ejecuta cada celda.



```python
from IPython.display import Markdown, display

from stock_toolkit.data.extractor import DataExtractor
from stock_toolkit.models.portfolio import Portfolio

# Exploración básica: composición y métricas descriptivas
extractor = DataExtractor.from_config()
series = extractor.fetch_many(["AAPL", "MSFT"], provider="yahoo", start_date="2022-01-01", end_date="2024-12-31")

portfolio = Portfolio.from_equal_weights(series)
report_md = portfolio.report(
    include_metrics=True,
    include_warnings=True,
    include_montecarlo=False,  # sin sección de Monte Carlo en este notebook
)
display(Markdown(report_md))

```
